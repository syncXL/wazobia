```python
import pandas as pd
from datasets import load_dataset, Features, Value
```

```python
DATASET_REPO = "nolimitsxl/wazobia-asr-data"
DATASET_DIR = "data/corpus=clearVoiceTWB"
errors = {
    "train" :  DATASET_DIR + "/split=train/language=hau",
    "val" :  DATASET_DIR + "/split=val/language=hau",
    "test" :  DATASET_DIR + "/split=test/language=hau"
}
```

```python
dict_count = {}
COLS_TO_REM = ["text","audio","snr_db", "noise_level_db", "speech_level_db", "speech_ratio", "spectral_bandwidth_hz"]

for e in errors:
    print(e)
    count = 0
    data = load_dataset(DATASET_REPO, data_dir=errors[e], streaming=True)
    data = data.remove_columns(COLS_TO_REM)
    data.cast_column("duration_sec", Value('float'))
    for row in data["train"]:
        if row["duration_sec"] is None:
            print("None Type encounterd")
            continue
        count += row["duration_sec"]
    dict_count[e] = count
```

```python
metric =  pd.read_csv("../data/metrics/metrics.tsv", delimiter="\t")
```

```python
grouped =  metric.groupby(by=["language", "corpus"],group_keys=["hours", "count"])
# grouped
```

```python
hours = grouped["hours"].sum()

totals = hours.groupby(level=0).sum()
totals.index = pd.MultiIndex.from_tuples(
    [(language, "Total") for language in totals.index],
    names=hours.index.names,
)

result = (
    pd.concat([hours, totals])
    .groupby(level=0, group_keys=False)
    .apply(lambda group: group.sort_values())
)

result
```
