```python
import pyarrow.dataset as pa_ds

# table = pa_ds.dataset(
#     "hf://buckets/nolimitsxl/wazobia",
#     partitioning="hive",
#     exclude_invalid_files=True
# )
import pyarrow.dataset as pa_ds
import polars as pl

def compute_metrics(parquet_root: str, group_cols=("corpus", "language", "split")) -> pl.DataFrame:
    metric_cols = [
        "snr_db", "noise_level_db", "speech_level_db",
        "speech_ratio", "duration_sec", "spectral_bandwidth_hz",
    ]

    ds = pa_ds.dataset(parquet_root, partitioning="hive", exclude_invalid_files=True)
    lf = pl.scan_pyarrow_dataset(ds)

    aggs = [pl.len().alias("count")]
    aggs.append((pl.col("duration_sec").sum() / 3600).alias("hours"))

    for col in metric_cols:
        aggs += [
            pl.col(col).mean().alias(f"{col}_mean"),
            pl.col(col).median().alias(f"{col}_median"),
            pl.col(col).std().alias(f"{col}_std"),
            pl.col(col).quantile(0.10).alias(f"{col}_p10"),
            pl.col(col).quantile(0.90).alias(f"{col}_p90"),
        ]

    return lf.group_by(list(group_cols)).agg(aggs).collect(streaming=True)

# compute_metrics()
ds = pa_ds.dataset("hf://datasets/nolimitsxl/wazobia-asr-data/data", partitioning="hive", exclude_invalid_files=True)
lf = pl.scan_pyarrow_dataset(ds)
```
